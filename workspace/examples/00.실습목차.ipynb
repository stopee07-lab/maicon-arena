# 3차 교육 영상 · ROS 2 실습

> 3차 교육 영상 20–30분의 기능을 재구성한 **로컬 실습본**입니다. 공식 원본 노트북이 아닙니다. `mil_msgs` 원본 정의가 없어 이 환경에서는 `sim_mil_interfaces`를 사용합니다. 실제 ROS 2로 시뮬레이터와 통신합니다.

영상 **20:00–30:00**에 나온 8개 예제와 같은 순서입니다. 아래 파일을 열고 위에서부터 **Shift+Enter**로 실행하세요.

**커널: Python 3 (ROS 2 · simulation)**

1. [개발 시작 · ROS 연결 확인](01.GettingStarted.ipynb)
2. [체력·배터리·경기 상태](02.State.ipynb)
3. [OLED 문자열 출력](03.Oled.ipynb)
4. [팬틸트 서비스](04.PanTilt.ipynb)
5. [주행 명령·위치](05.Driving_Odometry.ipynb)
6. [레이저 서비스·재발사 대기](06.Laser_Fire.ipynb)
7. [카메라 사진 한 장](07.Camera.ipynb)
8. [IMU·위치·정지 명령](08.Imu.ipynb)


## 시작 순서

1. 개발 화면에서 물리 경기장을 자유연습으로 켜고, 임무 주행 프로세스를 정지합니다.
2. **영상 실습 · ROS 2**를 엽니다. 이 창은 Ubuntu 22.04·ROS 2 Humble 커널을 사용합니다.
3. 01번에서 연결을 확인한 뒤 원하는 예제를 엽니다.
4. 주행은 `RUN_DRIVE`, 발사는 `RUN_FIRE`를 직접 True로 바꾼 경우에 실행합니다.
5. 셀을 중단하려면 위쪽 ■ 버튼 또는 Kernel → Interrupt Kernel을 사용합니다.

로봇 자체의 주행 알고리즘은 바꾸지 않았습니다. 각 노트북은 센서·명령을 직접 시험하는 별도 실습입니다.


## 영상과 현재 환경의 차이

| 항목 | 영상 | 현재 실습 |
|---|---|---|
| 실행 | 실제 UGV의 JupyterLab | Mac 안의 Ubuntu·ROS 2 JupyterLab |
| 로봇 | 지급 UGV | 3D 시뮬레이터 B-1 |
| 사용자 정의 메시지 | `mil_msgs` | `sim_mil_interfaces` |
| 카메라 | 최대 30FPS, 640×480 BGR | 640×480 BGR, 실제 주기는 실행 부하에 따름 |
| IMU 가속도 | 실물 IMU 값 | 미제공으로 표시 |

노트북을 저장하면 프로젝트 `simulator/workspace/examples`에 저장됩니다. 참가자 실행 코드 파일은 `simulator/workspace/team_node.py`입니다. 노트북 수정과 실행 중인 참가자 코드 버전 적용은 별개입니다.
